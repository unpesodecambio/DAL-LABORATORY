# delitos

```python
gdf_delitos = gpd.read_file(ruta_delitos)
```

```python
gdf_delitos.columns
```

```python
# --- 3. LÓGICA DE RECLASIFICACIÓN (MAPEO) ---
# Definimos las reglas según lo que me pediste
def clasificar_delitos(valor):
    valor = str(valor) # Por si hay nulos
    if valor in ['Robo a bancos', 'Robo a carga pesada',
                   'Robo a cuentahabientes', 'Robo a int de vehiculos',
                   'Robo a negocio', 'Robo a persona',
                   'Robo a vehiculos particulares', 'Robo casa habitacion',
                   'Robo de autopartes', 'Robo de motocicleta']:
        return "Bienes"
    elif valor in ['Feminicidio', 'Homicidio doloso','Lesiones dolosas']: 
        return "Corporal"
    elif valor in ['Abuso sexual infantil', 'Violacion','Violencia familiar']: 
        return "Hogar"
    else:
        return "Otra" # Para las que no entren en tus reglas

gdf_delitos['Categoría'] = gdf_delitos['delito'].apply(clasificar_delitos)

# Borrar columnas específicas
cols_a_borrar = [ 'x', 'y', 'colonia', 'municipio', 'bien_afectado', 'zona_geografica']

gdf_delitos = gdf_delitos.drop(columns=cols_a_borrar, errors='ignore')
```

```python
# Pads every value to a total length of 5 characters
gdf_delitos['clave_mun'] = gdf_delitos['clave_mun'].astype(str).str.zfill(3)
```

```python
# Guardar en la ruta especificada
gdf_delitos.to_file(ruta_salida, layer='1_D5C4V1_delitos', driver="GPKG")
```

# inundaciones

```python
gdf_inundaciones = gpd.read_file(ruta_inundaciones)
```

```python
# Borrar columnas específicas
cols_a_borrar = [ 'Superficie', 'Fuente']

gdf_inundaciones = gdf_inundaciones.drop(columns=cols_a_borrar, errors='ignore')
```

```python
# Guardar en la ruta especificada
gdf_inundaciones.to_file(ruta_salida, layer='1_D5C4V1_inundaciones', driver="GPKG")
```
