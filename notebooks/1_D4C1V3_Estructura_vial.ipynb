```python
import geopandas as gpd
import pandas as pd
import os
```

```python
# rutas alex
ruta_vias = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Vialidades\Tipo de Vía.gpkg"
ruta_salida = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_1\1_D4C1_Movilidad.gpkg"

CRS_MASTER = "MEXICO_ITRF_2008_LCC"

# # rutas aldo
# ruta_vias = r"C:\Users\aldov\OneDrive\Escritorio\DAL_LABORATY\data_raw\Vialidades\Tipo de Vía.gpkg"
# ruta_salida = r"C:\Users\aldov\OneDrive\Escritorio\DAL_LABORATY\outputs\1_D4C1_Movilidad.gpkg"
```

```python
# --- 2. CARGA DE DATOS ---
gdf_vial = gpd.read_file(ruta_vias)
```

```python
gdf_vial = gdf_vial[gdf_vial['Subclasifi'].isin(['V-S-Consolidar', 'V-S-Existente','V-F-Existente', 'V-P-Consolidar',
                                               'V-P-Existente', 'V-R-Consolidar', 'V-R-Existente'])].copy()

# --- 3. LÓGICA DE RECLASIFICACIÓN (MAPEO) ---
# Definimos las reglas según lo que me pediste
def clasificar_vias(valor):
    valor = str(valor) # Por si hay nulos
    if valor in ["Vialidad Primaria", "Vialidad primaria"]:
        return "Primaria"
    elif valor in ["Vialidad Secundaria", "Vialiadad Secundaria"]: # Incluye el error de dedo que mencionaste
        return "Secundaria"
    elif valor == "Vialidad Principal":
        return "Principal"
    elif valor == "Vialidad Regional":
        return "Regional"
    elif valor == "Vialidad Federal":
        return "Federal"
    else:
        return "Otra" # Para las que no entren en tus reglas

gdf_vial['Categoría'] = gdf_vial['Clasificac'].apply(clasificar_vias)

# --- 4. LIMPIEZA DE COLUMNAS ---
cols_a_borrar = ["Longitud_d", "Subclasifi", 'Clasificac']
gdf_vial = gdf_vial.drop(columns=cols_a_borrar, errors='ignore')

gdf_vial = gdf_vial.to_crs(CRS_MASTER)
```

```python
# Guardar en la ruta especificada
gdf_vial.to_file(ruta_salida, layer='1_D4C1V3_estructura_vial', driver="GPKG")
```
