# BLOQUE 01 – PERFIL DEMOGRÁFICO - Condicionantes
## Objetivo
Construir capa consolidada demográfica normalizada por unidad espacial.
## Inputs
- manzanas.gpkg
- censo_2020.csv
- conapo.xlsx
## Procesos
- Limpieza
- Homologación de CRS
- Join por CVEGEO

## Output
-1_D2C1V1._output_perfil_demografico.gpkg

```python
import pandas as pd
import geopandas as gpd
```

```python
# RUTAS ALEX
ruta_manzanas = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Manzanas\14m.shp"
ruta_conapo = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Rezago Social\conapo.xlsx"
ruta_censo = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Censos\RESAGEBURB_14CSV20.csv"

ruta_salida = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_1\1_D2C1_output_1_Poblacion.gpkg"

# # RUTAS DENI
# ruta_manzanas = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\manzanas.gpkg"
# ruta_conapo = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\conapo.xlsx"
# ruta_censo = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\censo\ageb_mza_urbana_14_cpv2020\conjunto_de_datos\conjunto_de_datos_ageb_urbana_14_cpv2020.csv"
```

# Censo

```python
# 2. LEER INFORMACIÓN CENSAL (Aplicando tu fix de dtype=str + encoding para el error)
manzanas = gpd.read_file(ruta_manzanas)
censo = pd.read_csv(ruta_censo, dtype={'ENTIDAD': str, 'MUN': str, 'LOC': str, 'AGEB': str, 'MZA': str})
```

```python
# Lista de columnas de población solicitadas
columnas_censo = ['PCON_DISC', 'PCDISC_MOT', 'PCDISC_VIS', 'PCDISC_LENG', 'PCDISC_AUD', 'PCDISC_MOT2', 'PCDISC_MEN']

# Crear clave de 16 dígitos (CVEGEO)
censo['CVEGEO'] = (
    censo['ENTIDAD'].str.zfill(2) + 
    censo['MUN'].str.zfill(3) + 
    censo['LOC'].str.zfill(4) + 
    censo['AGEB'].str.zfill(4) + 
    censo['MZA'].str.zfill(3)
)

# 3. CARGAR MAPA Y FILTRAR
manzanas['CVEGEO'] = manzanas['CVEGEO'].astype(str).str.strip()

# 4. UNIÓN FINAL (Solo Censo)
# Usamos how='left' para mantener todos los polígonos del mapa
D2C1V1 = manzanas.merge(censo[['CVEGEO'] + columnas_censo], on='CVEGEO', how='left')
```

```python
# Guardar
D2C1V1.to_file(ruta_salida, layer='1_D2C1V2_Condicionantes', driver='GPKG')

print("--- PROCESO TERMINADO ---")
print(f"El archivo D2C1V2.gpkg se generó con {len(D2C1V1)} manzanas y solo información del Censo.")
```

# Conapo

```python
conapo = pd.read_excel(ruta_conapo)
manzanas = gpd.read_file(ruta_salida, layer='1_D2C1V2_Condicionantes')
```

```python
conapo_clean = conapo[['Clave de la AGEB', 'Grado de Rezago Social']].copy()

conapo_clean = conapo_clean.rename(columns={'Clave de la AGEB': 'CVEAGEB', 'Grado de Rezago Social': 'REZAGO'})

# Crear clave de 13 dígitos (CVEAGEB)
manzanas['CVEAGEB'] = (
    manzanas['CVE_ENT'].str.zfill(2) + 
    manzanas['CVE_MUN'].str.zfill(3) + 
    manzanas['CVE_LOC'].str.zfill(4) + 
    manzanas['CVE_AGEB'].str.zfill(4)
)

# UNIONES
D2C1V1 = manzanas.merge(conapo_clean, on='CVEAGEB', how='left')
```

```python
# Guardar
D2C1V1.to_file(ruta_salida, layer='1_D2C1V2_Condicionantes', driver='GPKG')

print(f"Éxito")
```
