```python
import pandas as pd
import geopandas as gpd
import numpy as np
```

```python
ruta_economico = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_1\1_D3C1_Económico.gpkg"
ruta_equipamientos = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_1\1_D3C2_Equipamientos.gpkg"

ruta_salida = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_2\2_Dimensiones.gpkg"
```

# Carga de datos

```python
gpd.list_layers(ruta_equipamientos)
```

```python
ue = gpd.read_file(ruta_economico, layer = "1_D3C1V1_Unidades_económicas")
# ue_mzn = gpd.read_file(ruta_economico, layer = "1_D3C1V1_Unidades_Economicas_mzn")
eq = gpd.read_file(ruta_equipamientos, layer = "1_D3C2V1_equipamientos")
# eq_mz = gpd.read_file(ruta_equipamientos, layer = "1_D3C2V1_equipamientos_mzn")
```

```python
ue.columns
```

```python
eq.columns
```

# filtrado estado y municipios

```python
# EXTRAER JALISCO
ue_ent = ue[ue['cve_ent'] == '14'].copy()
eq_ent = eq[eq['CVE_ENT'] == '14'].copy()

# Lista de municipios AMG
municipios = ['002','039', '044', '051','070','097', '098', '101', '120', '124']

# Filtrar el DataFrame solo con los municipios de interés
ue_mun = ue_ent[
    (ue_ent['cve_mun'].isin(municipios)) & 
    (ue_ent['cve_ent'] == '14')
]

eq_mun = eq_ent[
    (eq_ent['CVE_MUN'].isin(municipios)) & 
    (eq_ent['CVE_ENT'] == '14')
]
```

# Extracción de elementos de interés

```python
actividades = ue_mun['nombre_act'].unique()

# buscar unidades económicas de interés
search_term = 'físi'
filtered_unique = [val for val in actividades if search_term in val]
filtered_unique
```

```python
actividades_interes = [
    'Consultorios del sector privado de audiología y de terapia ocupacional, física y del lenguaje',
     'Consultorios del sector público de audiología y de terapia ocupacional, física y del lenguaje']

ue_mun = ue_mun[ue_mun['nombre_act'].isin(actividades_interes)].copy()
```

```python
eq_mun['GEOGRAFICO'].unique()
```

```python
equipamientos = eq_mun[eq_mun['GEOGRAFICO'] == 'Centro de Asistencia Médica']['TIPO'].unique()
equipamientos
```

```python
eq_mun = eq_mun[(eq_mun['GEOGRAFICO'] == 'Centro de Asistencia Médica') & (eq_mun['TIPO'] == 'Centro de Rehabilitación')]
```

# Limpieza asteriscos y n/d

```python
# Reemplazar * con 1
def replace(dataframe):
    return dataframe.replace('*', '1')

# Limpiar *
ue_mun = replace(ue_mun)
eq_mun = replace(eq_mun)

# Reemplazar N/D
ue_mun = ue_mun.replace('N/D', np.nan)
eq_mun = eq_mun.replace('N/D', np.nan)

# Actualizar valores sin información de "null" a 0
ue_mun = ue_mun.fillna(0)
eq_mun = eq_mun.fillna(0)
```

# Guardado 

```python
ue_mun.to_file(ruta_salida, layer='2_D3C1_competencia_ue', driver='GPKG')
eq_mun.to_file(ruta_salida, layer='2_D3C2_competencia_eq', driver='GPKG')
```
