```python
import geopandas as gpd
import pandas as pd
import fiona
from pathlib import Path
import re
```

```python
# GEOPACKAGES A DOCUMENTAR

archivos = [
    r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_2\2_Dimensiones.gpkg",
    r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_3\3_Grid_ponderado.gpkg"
    # Agrega aquí los demás
]

registros = []

for archivo in archivos:

    nombre_archivo = Path(archivo).name

    for layer in fiona.listlayers(archivo):

        etapa = layer.split("_")[0]

        m = re.search(r"(D\d+(?:C\d+)?)", layer)
        dimension = m.group(1) if m else ""

        capa = re.sub(r"^\d+_", "", layer)
        capa = re.sub(r"^D\d+(?:C\d+)?_?", "", capa)

        gdf = gpd.read_file(archivo, layer=layer, rows=1)

        for campo, tipo in gdf.dtypes.items():

            if campo == "geometry":
                continue

            registros.append({
                "Etapa": etapa,
                "Dimensión": dimension,
                "Capa": capa,
                "Campo": campo,
                "Tipo": str(tipo),
                "Descripción": "",
                "Archivo": nombre_archivo
            })

diccionario = pd.DataFrame(registros)

diccionario = diccionario.sort_values(
    ["Etapa", "Dimensión", "Capa", "Campo"]
).reset_index(drop=True)

diccionario.to_csv(
    r"C:\Users\alflo\Desktop\DAL_LAB\outputs\Diccionario_Indicadores.csv",
    index=False
)

print("Diccionario generado.")
```
