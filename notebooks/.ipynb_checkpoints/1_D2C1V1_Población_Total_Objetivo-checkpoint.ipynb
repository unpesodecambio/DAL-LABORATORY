```python
# RUTAS ALEX
ruta_manzanas = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Manzanas\14m.shp"
ruta_censo = r"C:\Users\alflo\Desktop\DAL_LAB\data_raw\Censos\RESAGEBURB_14CSV20.csv"

ruta_salida = r"C:\Users\alflo\Desktop\DAL_LAB\outputs\etapa_1\1_D2C1_Poblacion.gpkg"

# # RUTAS DENI
# ruta_manzanas = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\manzanas.gpkg"
# ruta_conapo = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\conapo.xlsx"
# ruta_censo = r"C:\Users\denih\OneDrive\Escritorio\DAL_LAB\data_row\censo\ageb_mza_urbana_14_cpv2020\conjunto_de_datos\conjunto_de_datos_ageb_urbana_14_cpv2020.csv"
```

# Censo

```python
# 2. LEER INFORMACIÓN CENSAL (Aplicando tu fix de dtype=str + encoding para el error)
manzanas = gpd.read_file(ruta_manzanas)
censo = pd.read_csv(ruta_censo, dtype={'ENTIDAD': str, 'MUN': str, 'LOC': str, 'AGEB': str, 'MZA': str})
```

```python
# Lista de columnas de población solicitadas
columnas_censo = [
    'POBTOT', 'POBFEM', 'POBMAS', 'P_0A2', 'P_0A2_F', 'P_0A2_M', 'P_3YMAS', 
    'P_3YMAS_F', 'P_3YMAS_M', 'P_5YMAS', 'P_5YMAS_F', 'P_5YMAS_M', 'P_12YMAS', 
    'P_12YMAS_F', 'P_12YMAS_M', 'P_15YMAS', 'P_15YMAS_F', 'P_15YMAS_M', 'P_18YMAS', 
    'P_18YMAS_F', 'P_18YMAS_M', 'P_3A5', 'P_3A5_F', 'P_3A5_M', 'P_6A11', 
    'P_6A11_F', 'P_6A11_M', 'P_8A14', 'P_8A14_F', 'P_8A14_M', 'P_12A14', 
    'P_12A14_F', 'P_12A14_M', 'P_15A17', 'P_15A17_F', 'P_15A17_M', 'P_18A24', 
    'P_18A24_F', 'P_18A24_M', 'P_15A49_F', 'P_60YMAS', 'P_60YMAS_F', 'P_60YMAS_M', 
    'POB0_14', 'POB15_64', 'POB65_MAS'
]

# Crear clave de 16 dígitos (CVEGEO)
censo['CVEGEO'] = (
    censo['ENTIDAD'].str.zfill(2) + 
    censo['MUN'].str.zfill(3) + 
    censo['LOC'].str.zfill(4) + 
    censo['AGEB'].str.zfill(4) + 
    censo['MZA'].str.zfill(3)
)

# 3. CARGAR MAPA Y FILTRAR
manzanas['CVEGEO'] = manzanas['CVEGEO'].astype(str).str.strip()

# 4. UNIÓN FINAL (Solo Censo)
# Usamos how='left' para mantener todos los polígonos del mapa
D2C1V1 = manzanas.merge(censo[['CVEGEO'] + columnas_censo], on='CVEGEO', how='left')
```

```python
# Guardar
D2C1V1.to_file(ruta_salida, layer='1_D2C1V1_Pop_tot_obj', driver='GPKG')

print("--- PROCESO TERMINADO ---")
print(f"El archivo D2C1V1.gpkg se generó con {len(D2C1V1)} manzanas y solo información del Censo.")
```
